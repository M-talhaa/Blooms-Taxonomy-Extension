```python
# === Embeddings (MPNet) ===
# Fast path: vectors are first looked up in revision_pipeline/all_embeddings.npz
# (arrays "words" and "vectors", L2-normalized float32), mirroring retrain.py's
# _load_npz_cache and embed_vec. Only words missing from the npz are encoded with
# the SentenceTransformer model, which is created lazily on the first miss.
import numpy as np
from pathlib import Path

_st = None

def _get_st():
    global _st
    if _st is None:
        from sentence_transformers import SentenceTransformer
        _st = SentenceTransformer("sentence-transformers/all-mpnet-base-v2")
    return _st

_NPZ_CACHE = {}

def _load_npz_cache():
    fp = Path("revision_pipeline") / "all_embeddings.npz"
    if fp.exists():
        z = np.load(fp)
        _NPZ_CACHE.update(zip([str(w) for w in z["words"]], z["vectors"].astype(np.float32)))
        print(f"[EMBED] loaded {len(_NPZ_CACHE)} vectors from all_embeddings.npz")

_load_npz_cache()

_embed_cache = {}

def embed(word: str) -> np.ndarray:
    w = word.strip().lower()
    if w in _NPZ_CACHE:
        return _NPZ_CACHE[w]
    if w in _embed_cache:
        return _embed_cache[w]
    v = _get_st().encode([w], normalize_embeddings=True, convert_to_numpy=True)[0].astype(np.float32, copy=False)
    _embed_cache[w] = v
    return v
```

```python
# Core verb lists and their frequencies, loaded from the merged core.
# The core merges Sharunova et al. with Stanny 2016 and Newton et al. 2020.
# Crosswalk to original-taxonomy level codes: Knowledge/Remember to Kn,
# Understand/Comprehension to Cm, Apply/Application to Ap, Analyze/Analysis to An,
# Create/Synthesis to Sn, Evaluate/Evaluation to Ev.
# The merged core has 381 unique verbs and 607 verb-level assignments
# (the working copy used in the original submission had 358 verbs and 559 assignments;
# it is kept as revision_pipeline/core_notebook_original.csv).
# File: revision_pipeline/core_merged.csv, long format with columns
# verb, level, freq, sources; one row per verb-level pair.
import pandas as pd
from pathlib import Path

CORE_CSV = Path("revision_pipeline") / "core_merged.csv"
_core_long = pd.read_csv(CORE_CSV)

def _level_lists(code):
    # Words sorted by descending frequency (ties alphabetical).
    g = _core_long[_core_long["level"] == code]
    g = g.sort_values(["freq", "verb"], ascending=[False, True], kind="mergesort")
    return g["verb"].tolist(), [int(f) for f in g["freq"]]

knowledge_words, knowledge_weights = _level_lists("Kn")
comprehension_words, comprehension_weights = _level_lists("Cm")
application_words, application_weights = _level_lists("Ap")
analysis_words, analysis_weights = _level_lists("An")
synthesis_words, synthesis_weights = _level_lists("Sn")
evaluation_words, evaluation_weights = _level_lists("Ev")

wordlists = [knowledge_words, comprehension_words, application_words, analysis_words, synthesis_words, evaluation_words]
weights = [knowledge_weights, comprehension_weights, application_weights, analysis_weights, synthesis_weights, evaluation_weights]
namelist = ['knowledge', 'comprehension', 'application', 'analysis', 'synthesis', 'evaluation']
```

```python
LEVELS = ["Kn","Cm","Ap","An","Sn","Ev"]

wordlists = [
    knowledge_words,
    comprehension_words,
    application_words,
    analysis_words,
    synthesis_words,
    evaluation_words
]

weights = [
    knowledge_weights,
    comprehension_weights,
    application_weights,
    analysis_weights,
    synthesis_weights,
    evaluation_weights
]

# High/low split (freq threshold f0 = 10 as reported in the paper)
f0 = 10
wordlists_high = [[w for w,wt in zip(ws,wts) if wt >= f0] for ws,wts in zip(wordlists, weights)]
wordlists_low  = [[w for w,wt in zip(ws,wts) if wt <  f0] for ws,wts in zip(wordlists, weights)]

# Sanity checks (should match paper)
total_entries = sum(len(ws) for ws in wordlists)
unique_verbs  = len(set([w for ws in wordlists for w in ws]))
print("Core unique verbs:", unique_verbs, "| Core entries:", total_entries)  # expect 381 / 607
print("High entries per level:", [len(x) for x in wordlists_high])
print("Low entries per level :", [len(x) for x in wordlists_low])
```

```python
import numpy as np, pandas as pd
from IPython.display import display

LEVELS = ["Kn","Cm","Ap","An","Sn","Ev"]

def centroid_from_words(words):
    V = np.vstack([embed(w) for w in words])            # each already L2-normalized
    mu = V.mean(axis=0)
    mu = mu / np.linalg.norm(mu)                        # normalize centroid
    return mu

def cosine_dist(a, b):
    # centroids are normalized => cosine distance = 1 - dot
    return 1.0 - float(np.dot(a, b))

def pairwise_centroid_table(wordlists, title, out_csv=None):
    C = [centroid_from_words(ws) for ws in wordlists]
    D = np.zeros((6,6), dtype=float)
    for i in range(6):
        for j in range(6):
            D[i,j] = cosine_dist(C[i], C[j])

    df = pd.DataFrame(D, index=LEVELS, columns=LEVELS)
    print("\n" + title)
    display(df.round(3))
    if out_csv:
        df.to_csv(out_csv, index=True)
    return D, df

def mean_dn_from_pairwise(D):
    # d_n = mean of D(i, i+n) for i=1..6-n
    rows = []
    for n in range(1, 6):
        rows.append({"n": n, "d_n": float(np.mean(np.diag(D, k=n)))})
    return pd.DataFrame(rows)

# ---- Tables T1 to T3 (the paper's Tables 2 to 4): centroid cosine distances in MPNet space ----
D_all,  T1 = pairwise_centroid_table(wordlists,      "Table 1. Core (All) - centroid cosine distances", out_csv="T1_core_all.csv")
D_high, T2 = pairwise_centroid_table(wordlists_high, "Table 2. Core (High-frequency) - centroid cosine distances", out_csv="T2_core_high.csv")
D_low,  T3 = pairwise_centroid_table(wordlists_low,  "Table 3. Core (Low-frequency) - centroid cosine distances", out_csv="T3_core_low.csv")

print("\nMean centroid distances d_n (All core):")
display(mean_dn_from_pairwise(D_all).round(3))
print("Mean centroid distances d_n (High core):")
display(mean_dn_from_pairwise(D_high).round(3))
print("Mean centroid distances d_n (Low core):")
display(mean_dn_from_pairwise(D_low).round(3))


# Gap-trend Spearman correlation for the mean centroid distances d_n (n = 1..5).
# scipy's spearmanr p-value uses a t-approximation that is invalid at n = 5,
# so the exact permutation p-value is computed by enumerating all 5! = 120
# permutations of the five gap values.
from itertools import permutations
from scipy.stats import spearmanr

def gap_trend_spearman_exact(D):
    ns = np.arange(1, 6, dtype=float)
    dn = np.array([float(np.mean(np.diag(D, k=n))) for n in range(1, 6)])
    rho_obs = float(spearmanr(ns, dn)[0])
    rhos = np.array([float(spearmanr(ns, np.array(p))[0]) for p in permutations(dn)])
    p_one = float(np.mean(rhos >= rho_obs - 1e-12))               # one-sided: increasing trend
    p_two = float(np.mean(np.abs(rhos) >= abs(rho_obs) - 1e-12))  # two-sided
    return rho_obs, p_one, p_two

print("\nGap-trend Spearman (d_n vs n, n = 5; exact p over all 120 permutations):")
for _name, _D in [("All core", D_all), ("High core", D_high), ("Low core", D_low)]:
    _rho, _p1, _p2 = gap_trend_spearman_exact(_D)
    print(f"  {_name}: rho = {_rho:.3f}, one-sided exact p = {_p1:.4f}, two-sided exact p = {_p2:.4f}")
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Requires from the previous cell:
# D_all, D_high, D_low (each 6x6 numpy array)
# LEVELS = ["Kn","Cm","Ap","An","Sn","Ev"]

def plot_heatmap(D, title, fname=None):
    fig, ax = plt.subplots(figsize=(6, 5))
    im = ax.imshow(D, interpolation="nearest")
    ax.set_title(title)
    ax.set_xticks(range(len(LEVELS))); ax.set_yticks(range(len(LEVELS)))
    ax.set_xticklabels(LEVELS); ax.set_yticklabels(LEVELS)
    plt.setp(ax.get_xticklabels(), rotation=45, ha="right", rotation_mode="anchor")
    cbar = fig.colorbar(im, ax=ax)
    cbar.set_label("Cosine distance (1 - cos)")
    ax.set_xlabel("Cognitive level")
    ax.set_ylabel("Cognitive level")
    fig.tight_layout()
    if fname:
        fig.savefig(fname, dpi=300, bbox_inches="tight")
    plt.show()

def mean_dn(D):
    # d_n = average of nth superdiagonal
    return np.array([np.mean(np.diag(D, k=n)) for n in range(1, 6)], dtype=float)

def plot_dn_curves(D_all, D_high, D_low, fname=None):
    n = np.arange(1, 6)
    dn_all  = mean_dn(D_all)
    dn_high = mean_dn(D_high)
    dn_low  = mean_dn(D_low)

    fig, ax = plt.subplots(figsize=(6, 4))
    ax.plot(n, dn_all,  marker="o", label="All core")
    ax.plot(n, dn_high, marker="o", label="High-frequency core")
    ax.plot(n, dn_low,  marker="o", label="Low-frequency core")
    ax.set_xticks(n)
    ax.set_xlabel("Taxonomic gap n")
    ax.set_ylabel("Mean centroid distance d_n")
    ax.set_title("Mean centroid distance vs. taxonomic gap")
    ax.legend()
    fig.tight_layout()
    if fname:
        fig.savefig(fname, dpi=300, bbox_inches="tight")
    plt.show()

# --- Heatmaps (working copies; the paper's Fig. 1 is made by revision_pipeline/regen_core_figures.py) ---
plot_heatmap(D_all,  "Core (All) - centroid distance heatmap", fname="Fig_core_all_heatmap.png")
plot_heatmap(D_high, "Core (High-frequency) - centroid distance heatmap", fname="Fig_core_high_heatmap.png")
plot_heatmap(D_low,  "Core (Low-frequency) - centroid distance heatmap", fname="Fig_core_low_heatmap.png")

# --- Mean centroid distance curves (working copy; the paper's Fig. 2 is made by revision_pipeline/regen_core_figures.py) ---
plot_dn_curves(D_all, D_high, D_low, fname="Fig_mean_dn_curves.png")
```

```python
# === Core table (MPNet-consistent) ===
import pandas as pd
import numpy as np

LEVELS = ["Kn","Cm","Ap","An","Sn","Ev"]

wordlists = [
    knowledge_words,
    comprehension_words,
    application_words,
    analysis_words,
    synthesis_words,
    evaluation_words
]
weights = [
    knowledge_weights,
    comprehension_weights,
    application_weights,
    analysis_weights,
    synthesis_weights,
    evaluation_weights
]

def clean_word(w: str) -> str | None:
    if w is None:
        return None
    w = str(w).strip().lower()
    return w if w.isalpha() else None

rows = []
for lvl_name, ws, wts in zip(LEVELS, wordlists, weights):
    for w, wt in zip(ws, wts):
        cw = clean_word(w)
        if cw is None:
            continue
        rows.append((cw, lvl_name, int(wt)))

df_core = pd.DataFrame(rows, columns=["verb","level","weight"]).drop_duplicates()

# MPNet vectors (already normalized)
df_core["vector"] = df_core["verb"].map(embed)

print("Core verbs:", df_core["verb"].nunique(), "| rows:", len(df_core))
print(df_core.head(10))
```

```python
import numpy as np
import matplotlib.pyplot as plt

def dn_from_table(T):
    A = T.values if hasattr(T, "values") else np.array(T)
    return [float(np.mean(np.diag(A, k=n))) for n in range(1, 6)]

dn_all  = dn_from_table(T1)
dn_high = dn_from_table(T2)
dn_low  = dn_from_table(T3)

x = [1,2,3,4,5]

plt.figure(figsize=(6.2, 4.2), dpi=300)
plt.plot(x, dn_all,  marker="o", label="All core")
plt.plot(x, dn_high, marker="o", label="High-frequency core (freq ≥ 10)")
plt.plot(x, dn_low,  marker="o", label="Low-frequency core (freq < 10)")
plt.title("Mean centroid distance vs. taxonomic gap")
plt.xlabel("Taxonomic gap n")
plt.ylabel(r"Mean centroid distance $d_n$")
plt.xticks(x)
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.savefig("Fig_mean_centroid_distance_by_gap.png", dpi=300, bbox_inches="tight")
plt.show()
```
